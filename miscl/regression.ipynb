```python
import numpy as np

rng = np.random.default_rng(42)
n = 100_000

# True model: y = 2*x1 + 3*x2 + e
x1 = rng.normal(0, 1, n)
x2 = -0.8 * x1 + rng.normal(0, 0.6, n)   # x2 moves opposite to x1
e = rng.normal(0, 1, n)
y = 2 * x1 + 3 * x2 + e

def ols(X, y):
    X = np.column_stack([np.ones(len(y)), X])   # add intercept
    return np.linalg.lstsq(X, y, rcond=None)[0]

# Step 1: delta = slope from regressing x2 on x1
delta = ols(x1, x2)[1]
print(f"delta (x2 on x1 slope):        {delta:+.3f}")
print(f"check cov(x1,x2)/var(x1):      {np.cov(x1, x2)[0,1] / np.var(x1, ddof=1):+.3f}")

# Step 2: short regression (y on x1 only)
b_short = ols(x1, y)[1]
print(f"\nshort regression y ~ x1:       b1 = {b_short:+.3f}")
print(f"predicted b1 + b2*delta:       {2 + 3*delta:+.3f}")

# Step 3: long regression (y on x1 and x2)
b_long = ols(np.column_stack([x1, x2]), y)
print(f"\nlong regression y ~ x1 + x2:   b1 = {b_long[1]:+.3f}, b2 = {b_long[2]:+.3f}")
```
